```python
locations_df = spark.table("bronze.locations")
weathers_df = spark.table("bronze.weathers")

city_weather_df = (
    locations_df.alias("l")
    .join(weathers_df.alias("w"), on="ip", how="inner")
    .select(
        "ip",
        "l.city",
        "l.country",
        "l.lat",
        "l.long",
        "w.temperature_c",
        "w.windspeed_kmh",
        "w.winddirection_deg",
        "w.weathercode",
        "w.is_day",
        "w.weather_time",
        "w.ingested_at",
    )
)

# city_weather_df.display()
```

```python
city_weather_df.write.mode("overwrite").saveAsTable("silver.city_weather_records")
```
